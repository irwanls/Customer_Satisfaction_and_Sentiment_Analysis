# Cek Outlier

```python
numeric=[
    "Age",
    "Flight Distance",
    "Inflight wifi service",
    "Departure/Arrival time convenient",
    "Ease of Online booking",
    "Gate location",
    "Food and drink",
    "Online boarding",
    "Seat comfort",
    "Inflight entertainment",
    "On-board service",
    "Leg room service",
    "Baggage handling",
    "Checkin service",
    "Inflight service",
    "Cleanliness",
    "Departure Delay in Minutes",
    "Arrival Delay in Minutes"
]
for x in numeric:
    Q1 = df[x].quantile(0.25)
    Q3 = df[x].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    value_outlier = df[
        (df[x] < lower) |
        (df[x] > upper)
    ]
    print("Outlier",x,":")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Jumlah Outlier:",value_outlier[x].count(),"\n")
```

# EDA

## Univariat

### Kategorikal

```python
categorical_cols = ['Gender', 'Customer Type', 'Type of Travel', 'Class', 'satisfaction']

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

axes = axes.flatten()

for i, col in enumerate(categorical_cols):
    sns.countplot(
        data=df,
        x=col,
        ax=axes[i],
        order=df[col].value_counts().index
    )

    axes[i].set_title(f'Distribusi {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Jumlah User')
    axes[i].tick_params(axis='x', rotation=45)
    axes[i].margins(y=0.15)

    for container in axes[i].containers:
        axes[i].bar_label(
            container,
            fmt='%d',
            padding=3
        )

fig.delaxes(axes[5])

plt.tight_layout()
plt.show()
```

### Numerik

```python
numerical_cols = [
    'Age',
    'Flight Distance',
    'Departure Delay in Minutes',
    'Arrival Delay in Minutes'
]

fig, axes = plt.subplots(2, 2, figsize=(14, 9))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    sns.histplot(
        data=df,
        x=col,
        bins=30,
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(f'Distribusi {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Jumlah Pelanggan')

plt.tight_layout()
plt.show()
```

### Kolom Rating

```python
rating_cols = [
    'Inflight wifi service',
    'Departure/Arrival time convenient',
    'Ease of Online booking',
    'Gate location',
    'Food and drink',
    'Online boarding',
    'Seat comfort',
    'Inflight entertainment',
    'On-board service',
    'Leg room service',
    'Baggage handling',
    'Checkin service',
    'Inflight service',
    'Cleanliness'
]

fig, axes = plt.subplots(5, 3, figsize=(18, 20))
axes = axes.flatten()

for i, col in enumerate(rating_cols):
    sns.countplot(
        data=df,
        x=col,
        ax=axes[i],
        order=sorted(df[col].dropna().unique())
    )

    axes[i].set_title(f'Distribusi {col}')
    axes[i].set_xlabel('Rating')
    axes[i].set_ylabel('Jumlah Pelanggan')

# Hapus subplot yang tidak digunakan
for j in range(len(rating_cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

## Bivariat

### Encoding

```python

import pandas as pd

# Salin dataframe agar data asli tidak berubah
df_encoded = df.copy()

# Menghapus kolom ID karena bukan variabel analisis
df_encoded = df_encoded.drop(
    columns=['Unnamed: 0', 'id'],
    errors='ignore'
)

# Binary encoding
df_encoded['Gender'] = df_encoded['Gender'].map({
    'Male': 1,
    'Female': 0
})

df_encoded['Customer Type'] = df_encoded['Customer Type'].map({
    'Loyal Customer': 1,
    'disloyal Customer': 0
})

df_encoded['Type of Travel'] = df_encoded['Type of Travel'].map({
    'Business travel': 1,
    'Personal Travel': 0
})

df_encoded['satisfaction'] = df_encoded['satisfaction'].map({
    'satisfied': 1,
    'neutral or dissatisfied': 0
})

# One-Hot Encoding untuk Class
df_encoded = pd.get_dummies(
    df_encoded,
    columns=['Class'],
    dtype=int
)

# Melihat hasil encoding
df_encoded.head()
```

### Korelasi

```python
# Matriks korelasi
corr_matrix = df_encoded.corr(numeric_only=True)

plt.figure(figsize=(18, 14))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    linewidths=0.3
)

plt.title('Correlation Matrix Customer Satisfaction')
plt.tight_layout()
plt.show()
```

### Distribusi Rating per Class

```python
fig, axes = plt.subplots(5, 3, figsize=(20, 23))
axes = axes.flatten()

for i, col in enumerate(rating_cols):

    data = df[df[col].between(1, 5)]

    distribusi = pd.crosstab(
        data['Class'],
        data[col],
        normalize='index'
    ) * 100

    distribusi = distribusi.reindex(
        index=['Business', 'Eco', 'Eco Plus'],
        columns=[1, 2, 3, 4, 5],
        fill_value=0
    )

    distribusi.T.plot(
        kind='bar',
        ax=axes[i],
        width=0.8
    )

    # Data label setiap batang
    for container in axes[i].containers:
        axes[i].bar_label(
            container,
            fmt='%.1f%%',
            padding=3,
            fontsize=8
        )

    axes[i].set_title(col)
    axes[i].set_xlabel('Rating')
    axes[i].set_ylabel('Persentase (%)')
    axes[i].tick_params(axis='x', rotation=0)
    axes[i].margins(y=0.20)

    if i == 0:
        axes[i].legend(title='Class')
    else:
        axes[i].get_legend().remove()

for j in range(len(rating_cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```
